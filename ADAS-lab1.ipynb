```python
#PROGRAM 1:write a program to simulate an Automatic Headlight control System that turns the 
#headlights ON or OFF based on the ambient lightintensity.
import cv2
import numpy as np
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input, decode_predictions

#Load pretrained CNN
model=MobileNetV2(weights='imagenet')

#Read an image (camera sensor simulation)
image=cv2.imread("day.jpg")


#resize image
img=cv2.resize(image,(224,224))

#prepare image
x=np.expand_dims(img,axis=0)
x=preprocess_input(x)

#CNN prediction
predictions=model.predict(x)

print(decode_predictions(predictions,top=3)[0])

#Simulated brightness calculation
gray=cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
brightness=np.mean(gray)

print("Ambient Brightness:",brightness)

if brightness < 80:
    print("Headlights ON")
else:
    print("Headlights OFF")
```

```python
#Program 2:Write a program to simulate a parking Assistance System that display warning message based on
#the distance between the vehicle and an obstacle.
import cv2
from ultralytics import YOLO
#Load YOLOv8 moedel
model=YOLO("yolov8n.pt")
#Load parking image 
image=cv2.imread("parking.jpg")
#Run object detection
results=model(image)
#copy image
output=image.copy()
for result in results:
    boxes=result.boxes
    for box in boxes:
        #Bounding box coordinates
        x1,y1,x2,y2=map(int,box.xyxy[0])
        #Confidence
        confidence=float(box.conf[0])
        #class ID
        class_id=int(box.cls[0])
        #class name
        label=model.names[class_id]
        #width and Height
        width=x2-x1
        height=y2-y1
        #Bounding box area
        area=width*height
        #Estimate distance
        if area>120000:
            warning="STOP!"
            color=(0,0,255)
        elif area>60000:
            warning="BRAKE NOW"
            color=(0,140,255)
        elif area>25000:
            warning="SLOW DOWN"
            color=(0,255,255)
        else:
            warning="SAFE"
            color=(0,255,0)
        #Draw rectangle 
        cv2.rectangle(output,(x1,y1),(x2,y2),color,2)
        #object name
        cv2.putText(output,
                    f"{label} {confidence:.2f}",
                    (x1,y1-35),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.6,
                    color,
                    2)
        #Warning message
        cv2.putText(output,
                    warning,
                    (x1,y1-10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.8,
                    color,
                    2)
cv2.imshow("Parking Assistance System",output)
cv2.waitKey(0)
cv2.destroyAllWindows()        
```
