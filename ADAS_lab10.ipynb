```python
import numpy as np
desired_distance=40
actual_distance=30
our_speed=80
lead_speed=60
kp=0.5
ki=0.1
kd=0.2
accumulated_error=5
rate_of_error=2
error=desired_distance-actual_distance
P=kp*error
I=ki*accumulated_error
D=kd*rate_of_error
PID_output=P+I+D
our_speed=our_speed-PID_output
print("Distance Error =",error,"m")
print("P =",P)
print("I =",I)
print("D =",D)
print("PID Output =",PID_output)
print("Lead Vechicle Speed =",lead_speed,"km/h")
print("Our New Speed =",our_speed,"km/h")
```
